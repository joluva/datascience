```python
# importamos las librerias de analisis Pandas y Numpy
import pandas as pd
import numpy as np
```

```python
# importamos los archivos para depurar y analizar
# poblacion.csv - empleo.csv - esperanza_de_vida.csv - exportaciones.csv - hogares_vivienda_superficie.csv

# armo los data frame de cada archivo
poblacion = pd.read_csv('poblacion.csv', encoding='latin-1')

hogares_vivienda_superficie = pd.read_csv('hogares_viviendas_superficie.csv', encoding='latin-1')
```

```python
# creo una funcion para ir analizando las primeras cinco lineas de cada data frame
def primeras_ultimas_cinco_filas(lista_df):
  for nombre, df in lista_df.items():
    #print(f"DataFrame {i+1}: ", i)
    print(f"Procesando Data Frame:  {nombre}")
    print(f"Primeras cinco lineas")
    print("-" * 50)      # armo un separador visual entre cada DF
    print(df.head())     # muestro las primeras 5 lineas del data frame
    print("-" * 50)      # armo un separador visual entre cada DF
    print(f"Ultimas cinco lineas\n")
    print(df.tail())     # muestro las ultimas cinco filas del archivo
    print("_" * 50)      # armo un separador visual entre cada df
    print(f"\nCantidad de filas y columnas: ",df.shape)      # muestro la cantidad de filas y columnas del DF
    print(f"\nTipos de datos del archivo:\n ")
    print(df.dtypes)     # muestro los tipos de datos del DF
    print(f"\nPrincipales estadisticas descriptivas:\n")
    print(df.describe()) # visualizamos la estadistica del DF
    print("-" * 50)      # armo un separador visul entre cada DF

```

```python
# creo un diccionario con los nombres de los data frame
dataframe_dict = {
    'poblacion': poblacion,
    'hogares_vivienda_superficie': hogares_vivienda_superficie
}

# llamo a la funcion que muestra las primeras cinco filas
primeras_ultimas_cinco_filas(dataframe_dict)
```

```python
# filto el archivo de poblacion, elimino de la variable provincia el 'Total Pais'
poblacion_filtrado = poblacion[poblacion['provincia'] != 'Total País']
poblacion_filtrado
```

```python
# usamos la funcion merge para combinar las tablas poblacion_filtrado con hogares en un nuevo dataframe, que llamaremos
# poblacion_superficie, la combinacion se hace donde los valores de la columna provincia coincida en ambos dataframe
poblacion_superficie = pd.merge(poblacion_filtrado, hogares_vivienda_superficie, left_on=['provincia'], right_on=['provincia'], how='left') #how left union por izquierda
poblacion_superficie
```

```python
# calcular un nuevo campo de dencidad de poblacion
poblacion_superficie['densidad_poblacional'] = poblacion_superficie['poblacion_total'] / poblacion_superficie['superficie_km2']
poblacion_superficie
```

```python
# comprobamos que no haya datos nulos
poblacion_superficie.isnull().sum()
```

Creo columna (fuera_de_rango), que toma del dataframe poblacion_superficie datos de la columna 'densidad_poblacional', le resta la media (promedio) de todos los valores de densidad_de_poblacion, uso np.mean y lo divido por la desviación standard calculada aplicando np.std a densidad_poblacional. Esto standariza los valores, y los convierte en una medida llamada z-core, este valor indica cuantas medidas están por encima o por debajo de la media.

```python
# hago el calculo de los valores que estan por encima o debajo de la media
poblacion_superficie['fuera_de_rango'] = (poblacion_superficie['densidad_poblacional'] - np.mean(poblacion_superficie['densidad_poblacional'])) / np.std(poblacion_superficie['densidad_poblacional'])
poblacion_superficie
```

```python
# aplicamos la funcion describe sobre el nuevo campo calculado fuera_de_rango
poblacion_superficie['fuera_de_rango'].describe()
```

Calcular percentil de la columna densidad de poblacion
El percentil 99 es el valor por debajo del cual se encuentra el 99% de los datos. Es decir que solo el 1% de los datos son mayores a este valor

```python
# hago el calculo del percentil
p99 = np.percentile(poblacion_superficie['fuera_de_rango'], 99)
print(f"Calculo del percentil 99: ", p99)
```

```python
# muestro los registros mayores o iguales a p99
poblacion_superficie[poblacion_superficie['fuera_de_rango'] >= p99]
```

*   La densidad poblacional de Tierra del fuego es muy baja
*   Esta provincia esta por debajo del percentil 99 calculado



Analizamos la variable superficie_km2 de la tabla poblacion_superficie

```python
# busco la media en dentro de la variable superficie_km2 usando np.mean y generado una nueva variable rango_medio_km2
poblacion_superficie['rango_medio_km2'] = np.mean(poblacion_superficie['superficie_km2'])
poblacion_superficie
```

```python
# aplico la funcion describe sobre la variable rango_medio_km2
poblacion_superficie['superficie_km2'].describe()
```

```python
# hago el calculo del percentil 99 sobre rango_medio_km2
p99_rango_medio_km2 = np.percentile(poblacion_superficie['superficie_km2'], 99)
print(f"Calculo del percentil 99: ", p99_rango_medio_km2)
```
