## **Aplico herramientas de visualización de Python utilizando Matplotlib y Seaborn**

```python
# importo las librerias correspondientes
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# armo los dateframe correspondientes
#hogares = pd.read_csv('hogares_viviendas_superficie.csv', encoding='latin1')
poblacion = pd.read_csv('poblacion.csv', encoding='latin1')
esperanza = pd.read_csv('esperanza_de_vida.csv', encoding='latin1')
exportaciones = pd.read_csv('exportaciones.csv', encoding='latin1')
#empleo = pd.read_csv('empleo.csv', encoding='latin1')
```

```python
# muestro las primeras y últimas 5 lineas de los df
#print('Primeras 5 lineas: hogares\n', hogares.head(), '\n')
print('Primeras 5 lineas: poblacion\n', poblacion.head(), '\n')
print('Primeras 5 lineas: esperanza_vida\n', esperanza.head(), '\n')
print('Primeras 5 lineas: exportaciones\n', exportaciones.head(), '\n')
#print('Primeras 5 lineas: empleo\n', empleo.head())

print('--' * 60)

#print('Ultimas 5 lineas: hogares\n', hogares.tail(), '\n')
print('Ultimas 5 lineas: poblacion\n', poblacion.tail(), '\n')
print('Ultimas 5 lineas: esperanza_vida\n', esperanza.tail())
print('Ultimas 5 lineas: exportaciones\n', exportaciones.tail(), '\n')
#print('Ultimas 5 lineas: empleo\n', empleo.tail())
```

```python
# filtro de la base poblacion 'Total Pais'
poblacion_filtrado = poblacion[poblacion['provincia'] == 'Buenos Aires']
poblacion_filtrado
```

```python
# armo un subgrafico con un histograma y un boxplot
chart = plt.figure(figsize=[12,2.5])
# uso la funcion subplot( fila, columna, orden del grafico)

# grafico 1: boxplot de poblacion total pais
plt.subplot(121)
plt.boxplot(poblacion_filtrado["poblacion_total"], vert=False)
plt.title("Poblacion Total de Buenos Aires. Promedio: " + str(round(poblacion_filtrado["poblacion_total"].mean(),1)))

# subgrafico 2: histograma de poblacion total pais
plt.subplot(122)
sns.histplot(data = poblacion_filtrado, x="poblacion_total", bins=10, color = "blue",kde = True)
plt.axvline(poblacion_filtrado["poblacion_total"].mean(), color='red', linestyle='--')
plt.xlabel("Poblacion Total")
plt.ylabel("Frecuencia")
plt.title("Poblacion Total de Buenos Aires. Promedio: " + str(round(poblacion_filtrado["poblacion_total"].mean(),1)))

#Guardo el grafico en un png
chart.savefig("histograma_poblacion.png")
```

```python
# filtro de la base esperanza por anio de todas las provincias
esperanza_2020 = esperanza[esperanza['anio'] == 2020]
esperanza_2020
```

```python
# diagrama de dispesion

# Podemos cambiar el tamaño del grafico
charScat=plt.figure(figsize=[10,10])

# Definimos el tipo de grafico y las variables
plt.scatter(esperanza_2020["mujeres"], esperanza_2020["varones"], alpha=0.6, s=50, c='green')

# definimos los titulos de los ejes
plt.xlabel("mujeres"); plt.ylabel("varones")

# titulo del grafico
plt.title("Esperanza de Vida")

# Agrego lineas cuadriculada
plt.grid()

charScat.savefig("scatterplot_esperanza_vida.png")
```

```python
# filtro la base de exportaciones tomando los datos de total

exportaciones_filtrado = exportaciones[(exportaciones['rubro'] == 'Total') & (exportaciones['provincia'] != 'Total País')]
exportaciones_filtrado
```

```python
# convierto los valores de las columnas anio y value a numerico

exportaciones_filtrado['anio'] = exportaciones_filtrado['anio'].astype(int)
exportaciones_filtrado['value'] = exportaciones_filtrado['value'].astype(float)
```

```python
# filtro por provincia
#exportaciones_2020 = exportaciones_filtrado[exportaciones_filtrado['anio'] == 2020]

exportaciones_filtrado=exportaciones_filtrado[(exportaciones_filtrado["provincia"] == "Buenos Aires")]

charBar=plt.figure(figsize=[20,4.5])
exportaciones_filtrado.plot(kind='bar', x='anio', y='value')
plt.title("Exportaciones de Buenos Aires")
plt.show()

charBar.savefig("barplot_exportaciones.png")
```
