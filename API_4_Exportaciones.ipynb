### IMPORTO LIBRERIAS NECESARIAS

```python
# importa las librerias pandas, numpy, matplotlib, seaborn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

### PREPARO LOS DATASET

```python
# preparo los dataset
exportaciones = pd.read_csv('exportaciones.csv', encoding='latin1')
exportaciones
```

### MUESTRO DATOS DE INFORME BASICOS

```python
# muestro los premeros datos relevante
print('Base: exportaciones \n')
print(exportaciones.head())
print('\n', exportaciones.tail(),'\n')
print('\n', exportaciones.describe(),'\n')
```

### MODIFICO DATOS FALTANTES Y CONVIERTO A NUMERO

```python
exportaciones['value'] = exportaciones['value'].replace('-', np.nan)
```

```python
exportaciones['value'] = exportaciones['value'].astype(float)
```

```python
exportaciones_total = exportaciones[(exportaciones['rubro']=='Total') &
                                    (exportaciones['provincia'] != 'Total País') &
                                    (exportaciones['anio']==2020)]
exportaciones_total.head()
#exportaciones_total
```

```python
exportaciones.describe()
```

```python
# armo un bloxplot para detectar Outliers
plt.figure(figsize=(4, 4))
plt.boxplot(exportaciones_total["value"])
plt.title("Valor de Exportaciones - Presencia de Outliers")
plt.show()
```

```python

# genero código de barras de las exportaciones totales por provincias del 2020
exportaciones_total = exportaciones_total.sort_values(by='value', ascending=False)
plt.figure(figsize=(9, 3))
plt.bar(exportaciones_total['provincia'], exportaciones_total['value'], color='skyblue')
plt.xlabel('Provincia')
plt.ylabel('Value')
plt.title('Exportaciones Totales por Provincia - 2020')
plt.xticks(rotation=90)
plt.show()
```
