```python
# importamos las librerias necesarias pandas y numpy
import pandas as pd
import numpy as np
import os

#os.chdir("C:/TeclaLab/DataScience/VisualizacionYTableau/MOD-2/Python/")
```

```python
# importamos los archivos originales .CSV con los que trabajaremos
# datos macroeconomicos (PBI, inflacion, deuda, resultado comercial y fiscal)
macro = pd.read_csv("datos_macro.csv")

# tipos de cambios de brasil y de argentina(valor del dolar en reales y pesos argentina)
tc_br = pd.read_csv("tc_brasil.csv")
tc_ar = pd.read_csv("tc_argentina.csv")
```

```python
# importamos nuevamente el archivo pero sin las ultimas 3 filas
macro = pd.read_csv('datos_macro.csv', skipfooter=3, engine='python')

# filtramos las observaciones/filas que correspondan los paises y variables que queremos
macro_f = macro[(macro["Country"].isin(["Argentina","Chile","Brazil"])) &
                (macro["Subject Descriptor"] == "Gross domestic product, constant prices") &
                (macro["Units"] == "Percent change")]

macro_f
```

```python
# eliminamos las columnas que ya no necesitamos
macro_f = macro_f.drop(columns = ["WEO Country Code", "ISO", "WEO Subject Code", "Subject Descriptor",
                                  "Subject Notes", "Units", "Scale", "Country/Series-specific Notes",
                                  "Estimates Start After"])
macro_f
```

```python
# transponemos la tabla para tenerla en formato largo, dejando fijo el pais
# convirtiendo los años de columnas a filas, y definiendo la variable de valores como crecimiento_pbi
macro_tr = pd.melt(macro_f, id_vars = ["Country"], var_name="year", value_name="crecimiento_pbi")
macro_tr
```

```python
# cambiamos el nombre de country por pais
macro_tr = macro_tr.rename(columns = {"Country":"pais"})

macro_tr
```

```python
# ordenamos la tabla por pais y por año
macro_tr = macro_tr.sort_values(by = ["pais","year"])
macro_tr
```

```python
# tipos de variables
macro_tr.dtypes
```

```python
macro_tr
```

```python
# convertimos algunos de los campos a tipos de variables que nos sirvan: años como enteros, credimiento como float
macro_tr = macro_tr.astype({"pais": str, "year": int, "crecimiento_pbi": float})
macro_tr.dtypes
```

```python
# filtramos las observaciones anteriores a 2020, el ultimo año con datos completos
macro_tr = macro_tr[macro_tr["year"]<=2020]
macro_tr
```

```python
# calculamos las principales estadisticas descriptivas de la variable crecimiento
macro_tr['crecimiento_pbi'].describe()
```

```python
# concatenamos las tablas con el tipo de cambio de brasil y argentina
tc = pd.concat([tc_br, tc_ar])
tc
```

```python
# agregamos por pais, usando el promedio ( promedio de la tasa de crecimiento de pbi)
macro_ag = macro_tr.groupby(["pais"], as_index=False).agg({"crecimiento_pbi":"mean"})
macro_ag
```

```python
# agregacion con numpy
np.max(macro_tr["crecimiento_pbi"])
```

```python
# contamos la cantidad de valores faltantes en tipo de cambio

tc["tipo_de_cambio"].isnull().sum()
```

```python
# eliminamos las filas en las que aparezca al menos unmissing
tc_ar = tc_ar.dropna(how='any')
tc_br
```

```python
# eliminamos las filas en las que haya missing en la variable tipo_de_cambio
tc_ar = tc_ar.dropna(subset=["tipo_de_cambio"])
tc_ar
```

```python
# inconsistencia en la variable de testo
# en este caso reeplazamos Brasil (español), por Brazil (ingles)
tc_br['pais'] = np.where(tc_br['pais'] == 'Brasil', 'Brazil', tc_br['pais'])
tc_br
```

```python
# identificamos outliers (casos aparte)
tc_br["tc_estandarizado"] = (tc_br["tipo_de_cambio"] - np.mean(tc_br["tipo_de_cambio"]))/np.std(tc_br["tipo_de_cambio"])
tc_br
```

```python
# conservamos solo los cambios desde 1995
tc_br2 = tc_br[tc_br["year"]>=1995]
tc_br2
```

```python
# percentil 99 de los valores de tipo de cambio
p99 = np.percentile(tc_br2["tipo_de_cambio"],99)
p99
```
